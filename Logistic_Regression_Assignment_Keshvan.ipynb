# Theory

**Q1. What is Logistic Regression, and how does it differ from Linear Regression?**
  
**Ans:** Logistic Regression is a supervised learning algorithm used for **classification tasks**, predicting categorical outcomes (e.g., spam or not spam).  
- **Difference from Linear Regression:**  
  - **Linear Regression** predicts continuous values, using a straight-line relationship.  
  - **Logistic Regression** predicts probability values between 0 and 1, using a **sigmoid function** to model the output.  

**Q2. What is the mathematical equation of Logistic Regression?**

**Ans:** The mathematical form is:  
$$
[
P(Y=1 | X) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 X_1 + \beta_2 X_2 + ... + \beta_n X_n)}}
]
$$
where:  
- \( P(Y=1 | X) \) is the probability of the positive class.  
- \( \beta_0 \) is the intercept.  
- \( \beta_i \) are the feature coefficients.  

**Q3. Why do we use the Sigmoid function in Logistic Regression?**


**Ans:** The **sigmoid function** converts linear outputs into probabilities between **0 and 1**:  
$$
[
\sigma(z) = \frac{1}{1 + e^{-z}}
]
$$
- Ensures the output is interpretable as a probability.  
- Helps in decision-making by applying a threshold (e.g., **0.5** for binary classification).  


**Q4. What is the cost function of Logistic Regression?**

**Ans:** Instead of Mean Squared Error (MSE), Logistic Regression uses **Log Loss (Binary Cross-Entropy)**:  
$$
[
J(\theta) = - \frac{1}{m} \sum \left[ y \log (\hat{y}) + (1 - y) \log (1 - \hat{y}) \right]
]
$$
- Penalizes wrong predictions more when confidence is high.  
- Ensures stable gradient descent updates.  


**Q5. What is Regularization in Logistic Regression? Why is it needed?**

**Ans:** **Regularization** prevents **overfitting** by adding a penalty to large coefficients:  
- **L1 (Lasso) Regularization:** Adds **absolute** penalty $$(\sum |\beta_i|)$$, forcing some coefficients to **zero** (feature selection).  
- **L2 (Ridge) Regularization:** Adds **squared** penalty $$(\sum \beta_i^2)$$, reducing coefficient size but not setting them to zero.  
- Regularization is **needed** to improve **generalization** and prevent overfitting.  

**Q6. Explain the difference between Lasso, Ridge, and Elastic Net regression.**  
**Ans:** Lasso, Ridge, and Elastic Net are types of **regularization techniques** used to prevent **overfitting** by penalizing large coefficients in a model.  

**1. Ridge Regression (L2 Regularization)**
- Adds a **squared penalty** to large coefficients:  
  $$[
  J(\theta) = - \frac{1}{m} \sum \left[ y \log (\hat{y}) + (1 - y) \log (1 - \hat{y}) \right] + \lambda \sum \beta_i^2
  ]$$
- It **shrinks** the coefficients towards zero but **does not eliminate them**.  
- Helps in **handling multicollinearity** (highly correlated features).  
- Retains all features but reduces their impact.  

**2. Lasso Regression (L1 Regularization)**
- Adds an **absolute penalty** to large coefficients:  
  $$[
  J(\theta) = - \frac{1}{m} \sum \left[ y \log (\hat{y}) + (1 - y) \log (1 - \hat{y}) \right] + \lambda \sum |\beta_i|
  ]$$
- Unlike Ridge, **some coefficients become exactly zero**, performing **automatic feature selection**.  
- Useful when only a **few features** are important.  
- Can be unstable when features are highly correlated.  

**3. Elastic Net (Combination of L1 & L2)**
- **Combines** Ridge and Lasso penalties:  
  $$[
  J(\theta) = - \frac{1}{m} \sum \left[ y \log (\hat{y}) + (1 - y) \log (1 - \hat{y}) \right] + \lambda_1 \sum |\beta_i| + \lambda_2 \sum \beta_i^2
  ]$$
- Provides a balance between **feature selection (Lasso)** and **coefficient shrinkage (Ridge)**.  
- Works better than Lasso when **features are correlated**.  
- More stable in high-dimensional datasets.  


**Q7. When should we use Elastic Net instead of Lasso or Ridge?**  
**Ans:**  
Elastic Net is preferred when **Lasso and Ridge both have limitations**.  

**When to use Ridge (L2)?**  
- If **all features are important**, and you don’t want to eliminate any.  
- If features are **highly correlated** (Ridge keeps them).  

**When to use Lasso (L1)?**  
- If **feature selection** is needed (eliminates unimportant features).  
- Works well when there are **a few important features** and others should be ignored.  

**When to use Elastic Net?**  
- If there are **many correlated features** and Lasso struggles.  
- If Lasso is eliminating too many important features.  
- When **both regularization and feature selection** are needed.  

Elastic Net is a good default when working with **high-dimensional data** with **correlated features**.  


**Q8. What is the impact of the regularization parameter (λ) in Logistic Regression?**  

**Ans:**  
The regularization parameter **λ (lambda)** controls the strength of regularization in Logistic Regression. It helps prevent overfitting by penalizing large coefficients, thereby improving the model's generalization to unseen data.  

- When **λ is zero**, no regularization is applied, meaning the model relies entirely on the given data. This often leads to overfitting, where the model performs well on training data but poorly on new data.  

- When **λ is small**, there is a slight regularization effect. The model still captures the data patterns effectively but may still lean towards overfitting if the dataset is complex.  

- When **λ is large**, strong regularization is applied, forcing the model coefficients to shrink toward zero. This reduces overfitting but may lead to underfitting if λ is too high. In such cases, the model may fail to capture essential patterns in the data, leading to poor performance on both training and test sets.  

- If **λ is excessively high**, the regularization effect becomes extreme, causing the model to lose valuable information from the features. This results in very high bias and underfitting, where the model becomes too simplistic to make accurate predictions.  

To determine an optimal **λ**, techniques such as cross-validation are commonly used. The goal is to find a balance where the model avoids both overfitting and underfitting. Regularization tuning is an essential step in improving the robustness of Logistic Regression, especially when working with high-dimensional datasets.


**Q9. What are the key assumptions of Logistic Regression?**  
**Ans:**  
- **Independent observations:** No duplicated or related data points.  
- **Linearity in log-odds:** Features must be **linearly** related to log-odds, not raw output.  
- **No multicollinearity:** Highly correlated features should be avoided or regularized.  
- **Sufficiently large sample size:** Helps achieve stable predictions.  


**Q10. What are some alternatives to Logistic Regression for classification tasks?**

**Ans:** Logistic Regression is a powerful classifier but has limitations in handling **complex, non-linear data**. Some alternative models include:  

**1. Decision Trees**  
- Handles **non-linearity** in data.  
- Interpretable and easy to visualize.  
- Prone to **overfitting**, but can be controlled using pruning.  

**2. Random Forest**  
- Uses multiple decision trees (ensemble) for **better accuracy**.  
- Handles **missing values and feature interactions** well.  
- Can be computationally expensive.  

**3. Support Vector Machines (SVM)**  
- Works well with **high-dimensional data**.  
-Handles **non-linearity** using **kernel trick**.  
- Can be **slow on large datasets**.  

**Q11. What are Classification Evaluation Metrics?**  
**Ans:** Classification evaluation metrics measure model performance. Key metrics include:  
- **Accuracy:** Overall correctness of the model.  
- **Precision:** Ratio of correctly predicted positive observations to total predicted positives.  
- **Recall (Sensitivity):** Ratio of correctly predicted positives to actual positives.  
- **F1-Score:** Harmonic mean of precision and recall.  
- **ROC-AUC Score:** Measures model’s ability to distinguish between classes.  
- **Log Loss:** Measures how well probabilities match actual outcomes.  


**Q12. How does class imbalance affect Logistic Regression?**  
**Ans:** Class imbalance causes Logistic Regression to be **biased towards the majority class**, leading to:  
- Poor recall for the minority class.  
- Misleading accuracy metrics (high accuracy despite poor minority class prediction).  
- Solutions: Use **resampling techniques**, **class weights**, or **alternative evaluation metrics** like F1-score.

**Q12. What are Classification Evaluation Metrics?**  

**Ans:**  Classification evaluation metrics are used to assess the performance of a classification model. Some key metrics include:  

- **Accuracy**: The percentage of correctly classified instances. It works well when the classes are balanced but can be misleading with imbalanced datasets.  
- **Precision**: The proportion of true positive predictions among all positive predictions. It is useful in cases where false positives should be minimized.  
- **Recall (Sensitivity)**: The proportion of actual positive instances that were correctly classified. It is crucial when missing positives is costly.  
- **F1-Score**: The harmonic mean of precision and recall, balancing both metrics. It is useful when dealing with imbalanced data.  
- **ROC-AUC (Receiver Operating Characteristic - Area Under Curve)**: Measures how well the model separates classes by plotting the true positive rate vs. false positive rate.  
- **Log Loss**: Measures how well the predicted probabilities match the actual labels, with lower values indicating better calibration.  

Choosing the right metric depends on the problem, such as focusing on recall for medical diagnoses or precision for spam detection.  

**Q13. How does class imbalance affect Logistic Regression?**  

**Ans:** Class imbalance occurs when one class has significantly more samples than another. This can negatively impact Logistic Regression in several ways:  

- The model may **favor the majority class**, leading to high accuracy but poor performance on the minority class.  
- **Decision boundaries may be skewed**, making it harder to correctly classify minority class instances.  
- Metrics like **accuracy can be misleading**, as a model predicting all instances as the majority class may still show high accuracy.  
- The model may learn **biased coefficients**, reducing its ability to generalize.  

Solutions include using **balanced class weights**, **oversampling the minority class**, **undersampling the majority class**, or using **alternative evaluation metrics** like F1-score and AUC-ROC.  

**Q14. What is Hyperparameter Tuning in Logistic Regression?**  

**Ans:** Hyperparameter tuning involves selecting the best values for model parameters that are not learned from the data. In Logistic Regression, some key hyperparameters include:  

- **Regularization strength (λ or C)**: Controls the penalty for large coefficients, balancing overfitting and underfitting.  
- **Solver choice**: Different solvers (e.g., ‘liblinear’, ‘saga’, ‘lbfgs’) handle optimization differently.  
- **Class weights**: Adjusting weights for imbalanced datasets can improve model fairness.  

Techniques like **Grid Search**, **Random Search**, and **Bayesian Optimization** help find the best hyperparameter values by systematically testing different combinations.  

**Q16. How is Logistic Regression extended for multiclass classification?**  

**Ans:**  Logistic Regression is inherently binary but can be extended to multiclass problems using:  

- **One-vs-Rest (OvR)**: The model trains a separate binary classifier for each class, treating it as positive and the rest as negative. The class with the highest probability is chosen.  
- **Softmax (Multinomial) Regression**: Instead of predicting binary outcomes, it assigns probabilities to multiple classes using the Softmax function. This method directly models the probability distribution across all classes.  

Softmax is generally preferred for multiclass classification as it provides a more holistic approach, but OvR is computationally simpler and works well for many applications.

**Q17. What are the advantages and disadvantages of Logistic Regression?**  

**Ans:**  

**Advantages:**  
- Simple and easy to implement.  
- Computationally efficient.  
- Works well with linearly separable data.  
- Outputs interpretable probabilities.  
- Can handle large datasets.  

**Disadvantages:**  
- Assumes linear relationships between features and log-odds.  
- Struggles with highly correlated or non-linearly separable data.  
- Sensitive to outliers.  
- Performance degrades with high-dimensional datasets without regularization.  


**Q18. What are some use cases of Logistic Regression?**  

**Ans:**  Logistic Regression is widely used in various domains, including:  

- **Medical Diagnosis**: Predicting the likelihood of diseases based on patient data.  
- **Fraud Detection**: Identifying fraudulent transactions in banking.  
- **Marketing**: Customer churn prediction and lead conversion modeling.  
- **Spam Filtering**: Classifying emails as spam or not.  
- **Credit Scoring**: Assessing the risk of loan default.  
- **Sentiment Analysis**: Categorizing text reviews as positive or negative.  

It is preferred when interpretability is important and data is linearly separable.  


**Q19. What is the difference between Softmax Regression and Logistic Regression?**  

**Ans:**  
- **Logistic Regression** is a binary classifier that predicts probabilities for two classes using the sigmoid function.  
- **Softmax Regression** is a generalization of Logistic Regression for multiclass problems, using the softmax function to assign probabilities to multiple classes.  

While Logistic Regression classifies into two categories, Softmax Regression allows classification into three or more classes by considering all classes simultaneously.  


**Q20. How do we choose between One-vs-Rest (OvR) and Softmax for multiclass classification?**  

**Ans:**  The choice between **OvR** and **Softmax** depends on the dataset and computational efficiency:  

- **Use One-vs-Rest (OvR)** when the dataset is large, as it trains multiple binary classifiers separately and is computationally cheaper. It is also useful when classes are highly imbalanced.  
- **Use Softmax (Multinomial Regression)** when a holistic approach is needed, as it optimizes all classes together. It is generally preferred for balanced datasets and when capturing overall probability distributions is important.  

In general, Softmax provides better probability estimates for multiclass classification but may require more computational power than OvR.

```python
# Q1: Write a Python program that loads a dataset, splits it into training and testing sets, applies Logistic Regression, and prints the model accuracy

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score

# Load dataset
data = load_iris()
X, y = data.data, data.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Train Logistic Regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")
```

```python
# Q2: Write a Python program to apply L1 regularization (Lasso) on a dataset using LogisticRegression(penalty='l1') and print the model accuracy

model_l1 = LogisticRegression(penalty='l1', solver='liblinear', max_iter=1000)
model_l1.fit(X_train, y_train)

y_pred_l1 = model_l1.predict(X_test)
accuracy_l1 = accuracy_score(y_test, y_pred_l1)
print(f"L1 Regularization Accuracy: {accuracy_l1:.4f}")
```

```python
# Q3: Write a Python program to train Logistic Regression with L2 regularization (Ridge) using LogisticRegression(penalty='l2'). Print model accuracy and coefficients

model_l2 = LogisticRegression(penalty='l2', solver='lbfgs', max_iter=1000)
model_l2.fit(X_train, y_train)

y_pred_l2 = model_l2.predict(X_test)
accuracy_l2 = accuracy_score(y_test, y_pred_l2)
print(f"L2 Regularization Accuracy: {accuracy_l2:.4f}")
print("L2 Coefficients:", model_l2.coef_)
```

```python
# Q4: Write a Python program to train Logistic Regression with Elastic Net Regularization (penalty='elasticnet')

from sklearn.linear_model import SGDClassifier

model_elasticnet = SGDClassifier(loss='log_loss', penalty='elasticnet', l1_ratio=0.5, max_iter=1000)
model_elasticnet.fit(X_train, y_train)

y_pred_elastic = model_elasticnet.predict(X_test)
accuracy_elastic = accuracy_score(y_test, y_pred_elastic)
print(f"Elastic Net Regularization Accuracy: {accuracy_elastic:.4f}")
```

```python
# Q5: Write a Python program to train a Logistic Regression model for multiclass classification using multi_class='ovr'

model_ovr = LogisticRegression(multi_class='ovr', solver='lbfgs', max_iter=1000)
model_ovr.fit(X_train, y_train)

y_pred_ovr = model_ovr.predict(X_test)
accuracy_ovr = accuracy_score(y_test, y_pred_ovr)
print(f"OvR Multiclass Accuracy: {accuracy_ovr:.4f}")
```

```python
# Q6: Write a Python program to apply GridSearchCV to tune the hyperparameters (C and penalty) of Logistic Regression. Print the best parameters and accuracy

from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.01, 0.1, 1, 10],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear']
}

grid_search = GridSearchCV(LogisticRegression(max_iter=1000), param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)

print(f"Best Parameters: {grid_search.best_params_}")
print(f"Best Accuracy: {grid_search.best_score_:.4f}")
```

```python
# Q15: Write a Python program to train Logistic Regression on the Titanic dataset, handle missing values, and evaluate performance

import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Step 1: Load the Titanic dataset from seaborn
titanic = sns.load_dataset('titanic')

# Step 2: Select relevant features and target variable
features = ['pclass', 'sex', 'age', 'fare', 'embarked']
target = 'survived'
df = titanic[features + [target]]

# Step 3: Handle missing values
df.dropna(subset=['embarked'], inplace=True)  # Drop rows where 'embarked' is missing

# Step 4: Split dataset into features and target
X = df.drop(columns=[target])
y = df[target]

# Step 5: Define preprocessing steps
numeric_features = ['age', 'fare']
categorical_features = ['pclass', 'sex', 'embarked']

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),  # Fill missing values with mean
    ('scaler', StandardScaler())  # Standardize numeric features
])

categorical_transformer = Pipeline(steps=[
    ('encoder', OneHotEncoder(drop='first'))  # One-hot encode categorical features
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Step 6: Create pipeline with Logistic Regression
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])

# Step 7: Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1, stratify=y)

# Step 8: Train the model
model.fit(X_train, y_train)

# Step 9: Evaluate model performance
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.4f}")
```

```python
# Q16: Apply feature scaling (Standardization) before training a Logistic Regression model.
# Evaluate its accuracy and compare results with and without scaling.

# Step 1: Select relevant features and target variable (Continuation from Q15)
features = ['pclass', 'sex', 'age', 'fare', 'embarked']
target = 'survived'
df = titanic[features + [target]]

# Step 2: Handle missing values (Already done in Q15)
df.dropna(subset=['embarked'], inplace=True)  # Drop rows where 'embarked' is missing

# Step 3: Split dataset into features and target
X = df.drop(columns=[target])
y = df[target]

# Step 4: Define preprocessing for unscaled model (only encoding, no scaling)
numeric_features = ['age', 'fare']
categorical_features = ['pclass', 'sex', 'embarked']

# Preprocessor without scaling
preprocessor_no_scaling = ColumnTransformer(transformers=[
    ('num', SimpleImputer(strategy='mean'), numeric_features),  # Fill missing values with mean
    ('cat', OneHotEncoder(drop='first'), categorical_features)  # One-hot encode categorical features
])

# Preprocessor with scaling
preprocessor_with_scaling = ColumnTransformer(transformers=[
    ('num', Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='mean')),
        ('scaler', StandardScaler())  # Standardize numeric features
    ]), numeric_features),
    ('cat', OneHotEncoder(drop='first'), categorical_features)
])

# Step 5: Create pipelines for Logistic Regression (one without scaling, one with scaling)
model_no_scaling = Pipeline(steps=[
    ('preprocessor', preprocessor_no_scaling),
    ('classifier', LogisticRegression(max_iter=1000))
])

model_with_scaling = Pipeline(steps=[
    ('preprocessor', preprocessor_with_scaling),
    ('classifier', LogisticRegression(max_iter=1000))
])

# Step 6: Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1, stratify=y)

# Step 7: Train and evaluate the model without scaling
model_no_scaling.fit(X_train, y_train)
y_pred_no_scaling = model_no_scaling.predict(X_test)
accuracy_no_scaling = accuracy_score(y_test, y_pred_no_scaling)

# Step 8: Train and evaluate the model with scaling
model_with_scaling.fit(X_train, y_train)
y_pred_with_scaling = model_with_scaling.predict(X_test)
accuracy_with_scaling = accuracy_score(y_test, y_pred_with_scaling)

# Step 9: Print accuracy comparison
print(f"Model Accuracy without Scaling: {accuracy_no_scaling:.4f}")
print(f"Model Accuracy with Scaling: {accuracy_with_scaling:.4f}")
```

```python
# Q24: Train Logistic Regression and find the optimal C (regularization strength) using cross-validation

import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Load the Titanic dataset
df = sns.load_dataset('titanic')

# Drop rows with missing target variable (Survived)
df = df.dropna()

# Select features and target
X = df[['sex', 'embarked', 'pclass', 'age', 'fare', 'sibsp', 'parch']]
y = df['survived']

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Define categorical and numerical features
categorical_features = ['sex', 'embarked', 'pclass']
numeric_features = ['age', 'fare', 'sibsp', 'parch']

# Preprocessing: Handle missing values, scale numerical data, and encode categorical data
preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), numeric_features),
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
])

# Create pipeline with Logistic Regression
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])

# Define parameter grid for tuning C (regularization strength)
param_grid = {
    'classifier__C': [0.01, 0.1, 1, 10, 100]
}

# Use GridSearchCV for hyperparameter tuning
grid_search = GridSearchCV(model, param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)

# Get best hyperparameter and accuracy
best_C = grid_search.best_params_['classifier__C']
best_accuracy = grid_search.best_score_

# Print results
print(f"Best C (regularization strength): {best_C}")
print(f"Best Cross-Validated Accuracy: {best_accuracy:.4f}")

# Evaluate on test set
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)
test_accuracy = accuracy_score(y_test, y_pred)
print(f"Test Accuracy with Best C: {test_accuracy:.4f}")
```
