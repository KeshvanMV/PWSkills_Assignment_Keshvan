# Theory

**Q1: What is a Decision Tree, and how does it work?**  

**Ans:** A Decision Tree is a supervised learning algorithm used for classification and regression tasks. It works by splitting the dataset into smaller subsets based on feature values, forming a tree-like structure. Each internal node represents a feature-based decision, branches represent possible outcomes, and leaf nodes provide final predictions. The tree-building process continues recursively, aiming to minimize impurity and maximize information gain at each split.  


**Q2: What are impurity measures in Decision Trees?**  

**Ans:** Impurity measures quantify how mixed the class labels are within a node. A node is considered pure if it contains only one class. The two main impurity measures are **Gini Impurity**, which calculates the probability of incorrect classification, and **Entropy**, which measures the level of disorder or uncertainty in the data. These metrics guide the tree in selecting the best splits.  


**Q3: What is the mathematical formula for Gini Impurity?**  

**Ans:** The Gini Impurity is calculated using the formula:  

$$[
Gini = 1 - \sum p_i^2
]  $$

where $$( p_i )$$ represents the probability of a sample belonging to class \( i \). A lower Gini value indicates a purer node.  


**Q4: What is the mathematical formula for Entropy?**  

**Ans:** Entropy is computed as:  

$$[
Entropy = - \sum p_i \log_2 p_i
]$$  

where $$( p_i )$$ is the probability of class \( i \). Higher entropy indicates more disorder, while lower entropy signifies a more homogeneous node.  


**Q5: What is Information Gain, and how is it used in Decision Trees?**  

**Ans:** Information Gain (IG) measures the reduction in entropy after splitting a dataset on a feature. It is calculated as:  

$$[
IG = Entropy(Parent) - \sum \left( \frac{|S_v|}{|S|} \times Entropy(S_v) \right)
]$$  

where $$( S_v )$$ represents the subsets formed after splitting. The feature with the highest IG is chosen for splitting, ensuring the best separation of classes.  


**Q6: What is the difference between Gini Impurity and Entropy?**  

**Ans:** Both Gini Impurity and Entropy measure node impurity, but they differ in calculation and interpretation. **Gini Impurity** is computationally simpler and measures the probability of misclassification, while **Entropy** quantifies uncertainty and requires logarithmic computations. Gini tends to be faster, while Entropy is more sensitive to class distribution.  



**Q7: What is the mathematical explanation behind Decision Trees?**  

**Ans:** Decision Trees operate by selecting the feature that results in the greatest reduction in impurity. Mathematically, the algorithm evaluates each split using Gini Impurity or Entropy and chooses the split with the highest Information Gain. The recursive partitioning continues until a stopping condition is met, such as a minimum number of samples per node or maximum tree depth.  



**Q8: What is Pre-Pruning in Decision Trees?**  

**Ans:** Pre-Pruning is a technique used to stop tree growth early to prevent overfitting. It sets constraints such as **maximum depth, minimum samples per leaf, or minimum information gain** to halt unnecessary splits. This helps maintain model generalization but may risk underfitting if the tree is stopped too soon.  


**Q9: What is Post-Pruning in Decision Trees?**  

**Ans:** Post-Pruning involves growing the full tree first and then trimming unnecessary branches. It removes nodes that do not significantly improve performance, often using cross-validation to determine the optimal subtree. This method reduces overfitting while preserving important splits.  


**Q10: What is the difference between Pre-Pruning and Post-Pruning?**  

**Ans:** The main difference is **when pruning is applied**. Pre-Pruning stops tree growth early using conditions like depth limits, preventing over-complex trees. Post-Pruning builds a complete tree first and then removes weak splits, balancing accuracy and generalization. Post-Pruning is often more effective but requires additional computation.  


**Q11: What is a Decision Tree Regressor?**  

**Ans:** A Decision Tree Regressor is used for regression tasks, predicting continuous numerical values instead of class labels. Instead of impurity measures like Gini or Entropy, it minimizes errors using metrics like **Mean Squared Error (MSE) or Mean Absolute Error (MAE)**. It splits data to reduce variance, making stepwise predictions.  


**Q12: What are the advantages and disadvantages of Decision Trees?**  

**Ans:**  
**Advantages:**  
- Simple to understand and interpret.  
- Handles both numerical and categorical data.  
- Requires minimal data preprocessing.  
- Works well for small to medium-sized datasets.  

**Disadvantages:**  
- Prone to overfitting, requiring pruning or regularization.  
- Sensitive to noisy data.  
- Greedy splitting may lead to suboptimal trees.  


**Q13: How does a Decision Tree handle missing values?**  

**Ans:** Decision Trees handle missing values by using different strategies:  
- **Ignoring missing values** when choosing the best split.  
- **Imputing missing values** with the most frequent class or mean value.  
- **Surrogate splits**, which find alternative features to make decisions when a primary feature has missing values.  


**Q14: How does a Decision Tree handle categorical features?**  

**Ans:** Decision Trees handle categorical features by creating binary or multi-way splits based on unique category values. Common methods include:  
- **One-Hot Encoding:** Converting categories into binary features.  
- **Label Encoding:** Assigning numerical values to categories.  
- **Ordinal Splitting:** Arranging categories in a meaningful order.  

Some algorithms like **CART** use binary splits, while others like **ID3 and C4.5** can handle multi-way splits.  


**Q15: What are some real-world applications of Decision Trees?**  

**Ans:** Decision Trees are widely used in various domains, including:  
- **Medical Diagnosis:** Identifying diseases based on symptoms.  
- **Fraud Detection:** Analyzing transaction patterns to detect fraud.  
- **Customer Segmentation:** Classifying customers for marketing strategies.  
- **Credit Scoring:** Assessing loan eligibility based on financial history.  
- **Recommendation Systems:** Suggesting products based on user behavior.  

Decision Trees offer interpretability and efficiency, making them useful across industries.

```python
# Q16: Train a Decision Tree Classifier on the Iris dataset and print the model accuracy

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load dataset
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=1)

# Train Decision Tree Classifier
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

# Predict and evaluate accuracy
y_pred = clf.predict(X_test)
print("Model Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Q18: Train a Decision Tree Classifier using Entropy as the splitting criterion and print accuracy

clf_entropy = DecisionTreeClassifier(criterion="entropy")
clf_entropy.fit(X_train, y_train)

y_pred_entropy = clf_entropy.predict(X_test)
print("Model Accuracy using Entropy:", accuracy_score(y_test, y_pred_entropy))
```

```python
# Q23: Apply feature scaling before training a Decision Tree Classifier and compare its accuracy with unscaled data

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# Generate a synthetic dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=1)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Train a Decision Tree without feature scaling
clf_unscaled = DecisionTreeClassifier(random_state=1)
clf_unscaled.fit(X_train, y_train)
y_pred_unscaled = clf_unscaled.predict(X_test)
accuracy_unscaled = accuracy_score(y_test, y_pred_unscaled)

# Apply feature scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train a Decision Tree on scaled data
clf_scaled = DecisionTreeClassifier(random_state=1)
clf_scaled.fit(X_train_scaled, y_train)
y_pred_scaled = clf_scaled.predict(X_test_scaled)
accuracy_scaled = accuracy_score(y_test, y_pred_scaled)

# Print accuracies
print(f"Accuracy without Scaling: {accuracy_unscaled:.4f}")
print(f"Accuracy with Feature Scaling: {accuracy_scaled:.4f}")
```

```python
# Q24: Train a Decision Tree Classifier using One-vs-Rest (OvR) strategy for multiclass classification

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import accuracy_score

# Generate a synthetic multiclass dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=3, n_informative=6, random_state=1)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Train Decision Tree with One-vs-Rest strategy
clf_ovr = OneVsRestClassifier(DecisionTreeClassifier(random_state=1))
clf_ovr.fit(X_train, y_train)

# Make predictions
y_pred = clf_ovr.predict(X_test)

# Print accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"One-vs-Rest Decision Tree Accuracy: {accuracy:.4f}")
```

```python
# Q27: Train a Decision Tree Classifier, apply Cost Complexity Pruning (CCP), and visualize its effect on accuracy

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Generate dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=1)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Train Decision Tree without pruning
clf = DecisionTreeClassifier(random_state=1)
clf.fit(X_train, y_train)

# Get cost complexity pruning path
path = clf.cost_complexity_pruning_path(X_train, y_train)
ccp_alphas = path.ccp_alphas[:-1]  # Ignore the maximum alpha as it prunes everything

# Train models for different alpha values
train_scores = []
test_scores = []

for alpha in ccp_alphas:
    clf_pruned = DecisionTreeClassifier(ccp_alpha=alpha, random_state=1)
    clf_pruned.fit(X_train, y_train)

    y_train_pred = clf_pruned.predict(X_train)
    y_test_pred = clf_pruned.predict(X_test)

    train_scores.append(accuracy_score(y_train, y_train_pred))
    test_scores.append(accuracy_score(y_test, y_test_pred))

# Plot accuracy vs. alpha
plt.figure(figsize=(8, 5))
plt.plot(ccp_alphas, train_scores, label="Train Accuracy", marker='o')
plt.plot(ccp_alphas, test_scores, label="Test Accuracy", marker='o')
plt.xlabel("Alpha (Pruning Strength)")
plt.ylabel("Accuracy")
plt.title("Effect of Cost Complexity Pruning on Decision Tree Accuracy")
plt.legend()
plt.show()
```

```python
# Q28: Train a Decision Tree Classifier and evaluate its performance using Precision, Recall, and F1-Score

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import precision_score, recall_score, f1_score

# Generate dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=1)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Train Decision Tree Classifier
clf = DecisionTreeClassifier(random_state=1)
clf.fit(X_train, y_train)

# Predict on test data
y_pred = clf.predict(X_test)

# Calculate Precision, Recall, and F1-Score
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

# Print results
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
```

```python
# Q29: Train a Decision Tree Classifier and visualize the confusion matrix using seaborn

import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix

# Generate dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=1)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Train Decision Tree Classifier
clf = DecisionTreeClassifier(random_state=1)
clf.fit(X_train, y_train)

# Predict on test data
y_pred = clf.predict(X_test)

# Compute confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Visualize the confusion matrix
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=["Class 0", "Class 1"], yticklabels=["Class 0", "Class 1"])
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix")
plt.show()
```

```python
# Q30: Train a Decision Tree Classifier and use GridSearchCV to find the optimal values for max_depth and min_samples_split

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier

# Generate dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=1)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Define parameter grid
param_grid = {
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10]
}

# Perform Grid Search
clf = DecisionTreeClassifier(random_state=1)
grid_search = GridSearchCV(clf, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Best parameters and score
print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation Accuracy:", grid_search.best_score_)
```
